```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score
import pandas as pd
import numpy as np

# Fresh copy
model_data = master.copy()

# Target
target = "stockout_risk"

# Remove missing target
model_data = model_data.dropna(subset=[target])

# Separate X and y
X = model_data.drop(columns=[target])
y = model_data[target]

# Convert categorical columns using one-hot encoding
X = pd.get_dummies(X, drop_first=True)

# Replace infinite values
X = X.replace([np.inf, -np.inf], np.nan)

# Fill missing values
X = X.fillna(X.median(numeric_only=True))

# Make sure everything is numeric
X = X.apply(pd.to_numeric, errors="coerce")

# Fill any remaining NaN
X = X.fillna(0)

# Encode target if required
if y.dtype == "object":
    from sklearn.preprocessing import LabelEncoder
    
    le = LabelEncoder()
    y = le.fit_transform(y.astype(str))

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)

# Random Forest
rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

# Prediction
y_pred = rf_model.predict(X_test)

# Evaluation
print("\nAccuracy:", round(accuracy_score(y_test, y_pred), 4))

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

```python
# Create model-ready data from master

model_data = master.copy()

# Target
target = "stockout_risk"

# Convert date into useful numeric features
if "date" in model_data.columns:
    model_data["date"] = pd.to_datetime(model_data["date"])
    
    model_data["year"] = model_data["date"].dt.year
    model_data["month"] = model_data["date"].dt.month
    model_data["day"] = model_data["date"].dt.day
    model_data["day_of_week"] = model_data["date"].dt.dayofweek
    
    # Remove raw datetime column
    model_data = model_data.drop(columns=["date"])

# Separate X and y
X_final = model_data.drop(columns=[target])
y_final = model_data[target]

# One-hot encoding
X_final = pd.get_dummies(X_final, drop_first=True)

# Clean values
X_final = X_final.replace([np.inf, -np.inf], np.nan)
X_final = X_final.fillna(0)

print("Final feature shape:", X_final.shape)
print("DateTime columns:",
      X_final.select_dtypes(include=["datetime"]).columns.tolist())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

# Encode target if categorical
target_encoder_final = None

if y_final.dtype == "object" or str(y_final.dtype) == "category":
    target_encoder_final = LabelEncoder()
    y_final = target_encoder_final.fit_transform(
        y_final.astype(str)
    )

# Split
X_train_final, X_test_final, y_train_final, y_test_final = train_test_split(
    X_final,
    y_final,
    test_size=0.20,
    random_state=42,
    stratify=y_final
)

print("Final training data:", X_train_final.shape)
print("Final testing data:", X_test_final.shape)
```

```python
final_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)

final_model.fit(
    X_train_final,
    y_train_final
)

print("Final Random Forest model trained successfully.")
```

```python
def predict_stockout_risk(input_data):

    input_df = pd.DataFrame([input_data])

    # Handle date
    if "date" in input_df.columns:
        input_df["date"] = pd.to_datetime(input_df["date"])

        input_df["year"] = input_df["date"].dt.year
        input_df["month"] = input_df["date"].dt.month
        input_df["day"] = input_df["date"].dt.day
        input_df["day_of_week"] = input_df["date"].dt.dayofweek

        input_df = input_df.drop(columns=["date"])

    # One-hot encoding
    input_df = pd.get_dummies(input_df, drop_first=True)

    # Match training features exactly
    input_df = input_df.reindex(
        columns=X_final.columns,
        fill_value=0
    )

    # Clean
    input_df = input_df.replace(
        [np.inf, -np.inf],
        np.nan
    )

    input_df = input_df.fillna(0)

    # Prediction
    prediction = final_model.predict(input_df)[0]

    # Decode target
    if target_encoder_final is not None:
        prediction = target_encoder_final.inverse_transform(
            [prediction]
        )[0]

    return prediction
```

```python
def predict_stockout_probability(input_data):

    input_df = pd.DataFrame([input_data])

    # Handle date
    if "date" in input_df.columns:
        input_df["date"] = pd.to_datetime(input_df["date"])

        input_df["year"] = input_df["date"].dt.year
        input_df["month"] = input_df["date"].dt.month
        input_df["day"] = input_df["date"].dt.day
        input_df["day_of_week"] = input_df["date"].dt.dayofweek

        input_df = input_df.drop(columns=["date"])

    # Encoding
    input_df = pd.get_dummies(
        input_df,
        drop_first=True
    )

    # Match training columns
    input_df = input_df.reindex(
        columns=X_final.columns,
        fill_value=0
    )

    # Clean
    input_df = input_df.replace(
        [np.inf, -np.inf],
        np.nan
    )

    input_df = input_df.fillna(0)

    # Probability
    probabilities = final_model.predict_proba(input_df)[0]

    # Classes
    classes = final_model.classes_

    # Decode classes
    if target_encoder_final is not None:
        classes = target_encoder_final.inverse_transform(classes)

    result = pd.DataFrame({
        "Risk Level": classes,
        "Probability": probabilities
    })

    result["Probability"] = (
        result["Probability"] * 100
    ).round(2)

    return result.sort_values(
        "Probability",
        ascending=False
    )
```
